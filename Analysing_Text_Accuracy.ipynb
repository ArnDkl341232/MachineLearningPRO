```python
from rouge_score import rouge_scorer

# reference = "Машинне навчання - це галузь штучного інтелекту, яка дозволяє комп'ютерам вчитися на даних"
reference = "Machine learning - is a branch of artificial intelligence, that allows computer learn by data"

# candidate_1 = "Машинне навчання - це галузь штучного інтелекту, де комп'ютери вчиться на даних"
candidate_1 = "Machine learning - is a branch of artificial intelligence where computers learn from data"
# candidate_2 = "Машинне навчання - це галузь ШІ"
candidate_2 = "Machine learning - is a branch of AI"
# candidate_3 = "Сьогодні хмарно, буде дощ"
candidate_3 = "It's cloudy today, it will rain."

scorer = rouge_scorer.RougeScorer(['rouge1', 'rouge2', 'rougeL'], use_stemmer=True)

for i, candidate in enumerate([candidate_1, candidate_2, candidate_3], 1):
    score = scorer.score(reference, candidate)
    #print(f"{1}) {candidate}")

    print(f"\n--- Candidate {i} ---")
    scores = scorer.score(reference, candidate)
    #print(scores)

    for key, value in scores.items():
        print(f"{key}: precision={value.precision:.2f}, recall={value.recall:.2f}, fmeasure={value.fmeasure:.2f}")
```

#Perplexity

```python
!pip install transformers torch -q
```

```python
from transformers import GPT2LMHeadModel, GPT2TokenizerFast
import torch
model_name = "gpt2"
model = GPT2LMHeadModel.from_pretrained(model_name)
tokenizer = GPT2TokenizerFast.from_pretrained(model_name)
model.eval

def calculate_perplexity(text):
  encoding = tokenizer(text, return_tensors="pt")
  input_ids = encoding["input_ids"]
  #attention_mask = encoding["attention_mask"]
  with torch.no_grad():
    outputs = model(input_ids, labels=input_ids)
    # print(outputs)
    loss = outputs.loss
    perplexity = torch.exp(loss)
    return perplexity.item()
```

```python
calculate_perplexity("The cat is sitting on the carpet.")
```

```python
text1 = "The cat is sitting on the carpet."
text2 = "Colorless green ideas sleep furiously."
tetx3 = "The quick brown fox jumps over the lazy dog."
text4 = "table run quickly purple democracy banana."

for i, text in enumerate([text1, text2, tetx3, text4], start=1):
  ppl = calculate_perplexity(text)
  print(f"Perplexity: {ppl:.2f} | Text{i}: {text}")
```
